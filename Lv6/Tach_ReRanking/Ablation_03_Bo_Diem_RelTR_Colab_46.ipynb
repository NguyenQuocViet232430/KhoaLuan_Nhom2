```python
import shutil
from pathlib import Path

#base_dir = Path('/content/drive/MyDrive/DataKhoaLuan/Data/baseline_level6')
base_dir = Path('/content/drive/MyDrive/Data/baseline_level6')
print("🧹 Đang tự động quét và dọn các thư mục trùng lặp...")

for sub in ['_baseline', 'Tach_Dimersity', 'Tach_Sematic', 'Tach_ReRanking', '_test_protocol']:
    target_parent = base_dir / sub
    if not target_parent.exists():
        continue

    # Tìm tất cả các folder seed45
    seed_dirs = [p for p in target_parent.iterdir() if 'seed45' in p.name]
    if len(seed_dirs) > 1:
        print(f"⚠️ Phát hiện {len(seed_dirs)} thư mục seed45 trong {sub}:")

        # Đếm số file bên trong để giữ lại thư mục có dữ liệu thật
        def count_files(p):
            return sum(1 for _ in p.glob('**/*') if _.is_file()) if p.is_dir() else 0

        # Sắp xếp: thư mục nhiều file nhất xếp đầu tiên
        seed_dirs.sort(key=count_files, reverse=True)
        main_dir = seed_dirs[0]
        print(f"   -> GIỮ LẠI: {main_dir.name} ({count_files(main_dir)} files)")

        # Xóa các thư mục rỗng thừa
        for dup in seed_dirs[1:]:
            n = count_files(dup)
            if n == 0:
                print(f"   🗑️ Đã xóa thư mục rỗng: {dup.name}")
                shutil.rmtree(dup, ignore_errors=True)
            else:
                print(f"   ⚠️ Thư mục có {n} files, tạm giữ: {dup.name}")
    else:
        print(f"✅ {sub}: Bình thường (1 thư mục)")

print("\n✨ DỌN DẸP HOÀN TẤT!")
```

## 5. Sinh caption → chấm COCO chính thức → bảng so sánh
Có resume caption theo chunk và từng metric đã hoàn tất. Nếu Colab ngắt, chạy lại cùng seed/config. METEOR/SPICE dùng Java; metric thiếu/lỗi sẽ dừng, không thay bằng 0. File log ở thư mục kết quả. Không dùng thời gian toàn bước này làm Online FPS.

```python
"""Read-only Test identity diagnostics, deliberately outside hashed runner code."""
import json
from pathlib import Path


def identity_differences(saved, current, prefix=''):
    if isinstance(saved, dict) and isinstance(current, dict):
        rows = []
        for key in sorted(set(saved) | set(current)):
            name = f'{prefix}.{key}' if prefix else key
            if key not in saved or key not in current:
                rows.append(dict(field=name, saved_present=key in saved,
                                 current_present=key in current,
                                 saved=saved.get(key), current=current.get(key)))
            else:
                rows.extend(identity_differences(saved[key], current[key], name))
        return rows
    return [] if saved == current else [dict(field=prefix, saved=saved, current=current)]


def diagnose_test_lock(experiment, output_dir='/content'):
    path = Path(experiment.test_record_path)
    current = experiment.pre_reference_identity
    saved = json.loads(path.read_text(encoding='utf-8')) if path.is_file() else None
    report = dict(lock_path=str(path), lock_exists=path.is_file(),
                  saved_identity=None if saved is None else saved.get('identity'),
                  current_identity=current,
                  differences=[] if saved is None else identity_differences(saved.get('identity'), current),
                  saved_status=None if saved is None else saved.get('status'),
                  completed_variants=None if saved is None else saved.get('completed_variants'),
                  reference_sha256=None if saved is None else saved.get('reference_sha256'),
                  note='Read-only diagnosis; source lock, captions and metrics were not changed.')
    output = Path(output_dir)/f'lv6_test_lock_diagnostic_seed{experiment.seed}.json'
    output.parent.mkdir(parents=True, exist_ok=True)
    output.write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding='utf-8')
    print('Khóa Test:', path)
    for row in report['differences']:
        print(json.dumps(row, ensure_ascii=False, indent=2))
    if not report['differences']:
        print('Không thấy khác identity; giữ nguyên traceback để kiểm tra bước khác.')
    print('Báo cáo chẩn đoán:', output)
    print('Không xóa khóa Test, đổi SEED, hoặc sửa hash để bỏ qua lỗi.')
    return output

# fcntl chỉ khóa trong runtime này, không khóa xuyên nhiều máy Colab.
import fcntl
lock_path = Path('/tmp')/('l6_' + experiment.contract_id + '.lock')
with lock_path.open('w') as lock:
    fcntl.flock(lock, fcntl.LOCK_EX | fcntl.LOCK_NB)
    try:
        comparison_rows = experiment.run_ablation(VARIANT)
    except ValueError as exc:
        if 'Test đã khóa' in str(exc):
            try:
                diagnose_test_lock(experiment)
            except Exception as diagnosis_error:
                print('Không xuất được chẩn đoán:', diagnosis_error)
        raise
    finally:
        fcntl.flock(lock, fcntl.LOCK_UN)
```
